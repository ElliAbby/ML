Вектор параметров ww следует искать с помощью алгоритма стохастического градиентного спуска (SGD):

$wn=wn−1−η⋅\frac{∂Qk(w)}{∂w}$

То есть, псевдоградиент вычисляется не по одному образу k, а по образам в диапазоне [k; k+K) (всего K образов).

$L_i(w)=e^{−Mi}=e^{−w^T⋅x_i⋅y_i}$

$\frac{∂Li(w)}{∂w}=−e^{−w^T⋅x_i⋅y_i}⋅x_i^T⋅y_i$
