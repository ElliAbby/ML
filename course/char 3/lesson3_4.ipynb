```python
import numpy as np


def func(x):
    return 0.5 * x + 0.2 * x ** 2 - 0.05 * x ** 3 + 0.2 * np.sin(4 * x) - 2.5


def model(w, x):
    return w[0] + w[1] * x + w[2] * x ** 2 + w[3] * x ** 3


coord_x = np.arange(-4.0, 6.0, 0.1)

x_train = np.array([[_x**i for i in range(4)] for _x in coord_x]) # обучающая выборка
y_train = func(coord_x) # целевые выходные значения

# здесь продолжайте программу
# Решаем уравнение w = (X^T * X)^(-1) * X^T * y
w = np.linalg.inv(x_train.T @ x_train) @ x_train.T @ y_train
w = list(w)

# Вычисляем средний эмпирический риск
y_pred = np.array([model(w, x) for x in coord_x])
Q = np.mean((y_train - y_pred) ** 2)
print(f"Q = {Q}")
```

```python
# @title Решение задачи регрессии с помощью теоремы Байеса
import numpy as np

# Функции
f = lambda x: 0.5 * x + 0.2 * x ** 2 - 0.05 * x ** 3 + 0.2 * np.sin(4 * x) - 2.5
p_prod = lambda X, Y, w, D: - 1 / D * (Y - X @ w) @ X

# Данные
coord_x = np.arange(-4.0, 6.0, 0.1)
x_train = np.array([[_x**i for i in range(4)] for _x in coord_x])
y_train = f(coord_x)

# Параметры
N = 10000
w = np.array([.0, .0, .0, .0])
k = np.array([.1, .001, .0001, .00001])

# Вычисление дисперсии
D = np.var(coord_x) # по двум осям

# Вычисление начального качества:
Q = np.mean((x_train @ w - y_train)**2)
print('Q начальное:', Q)

for _ in range(N):
    grad = p_prod(x_train, y_train, w, D)
    w -= k * grad
    if _ % 1000 == 0:
        Q = np.mean((x_train @ w - y_train)**2)
        print(f'Итерация {_}, w: {w}, Q: {Q}')

# Вычисление конечного качества:
Q = np.mean((x_train @ w - y_train)**2)
print('Q конечное:', Q)
```
